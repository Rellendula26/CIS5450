```python
gmail = input("Enter your Gmail address: ")
password = getpass("Enter your Gmail App Password: ")
```

```python
if len(emails) != len(names):
    raise Exception("Emails and names lists are different lengths.")

server = smtplib.SMTP_SSL("smtp.gmail.com", 465)
server.login(gmail, password)

for email, name in zip(emails, names):

    personalized_body = body.replace("{{name}}", name)

    message = MIMEMultipart()
    message["From"] = gmail
    message["To"] = email
    message["Subject"] = subject

    message.attach(MIMEText(personalized_body, "plain"))

    try:
        server.sendmail(gmail, email, message.as_string())
        print(f"✅ Sent to Dr. {name}")
    except Exception as e:
        print(f"❌ Failed to send to Dr. {name}: {e}")

server.quit()

print("\nFinished!")
```
